## components
root node
leaf node
node's sample: how many training instances it applies to. 
node's value: how many training instances of each class this node applies to. 
node's gini: measure its impurity. 
## parameters 
max_depth hyperparameter (the default value is None, which means unlimited)

min_samples_split (the minimum number of samples a node must have before it can be split)

min_samples_leaf (the minimum number of samples a leaf node must have)


# Exercises
## What is the approximate depth of a Decision Tree trained (without restrictions) on a training set with 1 million instances?
the depth is around log2(m) where m = 1 million. This is about 20. 
## Is a node’s Gini impurity generally lower or greater than its parent’s? Is it generally lower/greater, or always lower/greater?
Generally lower (but only the weighted average impurity)
## If a Decision Tree is overfitting the training set, is it a good idea to try decreasing max_depth?
Yes, it is a good idea to regularize the model.
## If a Decision Tree is underfitting the training set, is it a good idea to try scaling the input features?
No, the decision trees is independent on the scaling.
