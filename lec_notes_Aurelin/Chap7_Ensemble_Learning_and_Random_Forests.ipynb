## Terminology
Ensemble: a group of predictors

Random Forest: ensemble of Decision Trees

Ensemble methods: bagging, boosting, stacking, random forest

## Bagging and Pasting
Use the same trainning algorithms but to train them in different subsets of the trainning set. 

Bagging: sampling is performed with replacement. 

Pasting: sampling is performed without replacement. 

Aggregation = statistical mode (classification).


## Random Patches and Random Subspaces
Random Patches method: Sampling both training instances and features. 

Random Subspaces: keeping all training instances but sampling features. 

## Random Forests
Feature importance: Scikit-Learn measures a feature’s importance by looking at how much the tree nodes that use that feature reduce impurity on average (across all trees in the forest)

## Boosting
Combine several weak learners into a strong learner.

AdaBoost: create new predictor to correct its predecessor when does underfitting (update the parameters)

Gradient boosting:  tries to fit the new predictor to the residual errors made by the previous predictor. XGBoost implement gradient boosting.

perform a model to perform the aggegation of predictors.
